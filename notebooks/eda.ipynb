```python
from scipy.stats import fisher_exact

val_norte_detrator = df[(df['customer_region'] == 'Norte') & (df['detrator'] == 1) & (df['delivery_delay_days'] == 0)].shape[0]
val_outros_detrator = df[(df['customer_region'] != 'Norte') & (df['detrator'] == 1) & (df['delivery_delay_days'] == 0)].shape[0] - val_norte_detrator
val_norte_total = df[(df['customer_region'] == 'Norte') & (df['delivery_delay_days'] == 0)].shape[0]
val_outros_total = df[(df['customer_region'] != 'Norte') & (df['delivery_delay_days'] == 0)].shape[0] - val_norte_total
```

```python
df_sem_atraso = df[df['delivery_delay_days'] == 0].copy()
df_sem_atraso['is_norte'] = df_sem_atraso['customer_region'] == 'Norte'
```

```python
pd.crosstab(df_sem_atraso['is_norte'], df_sem_atraso['detrator'])
```

```python
fisher_exact(pd.crosstab(df_sem_atraso['is_norte'], df_sem_atraso['detrator']))
```

```python
from scipy.stats import chi2_contingency

chi2_contingency(pd.crosstab(df_sem_atraso['is_norte'], df_sem_atraso['detrator']))
```
