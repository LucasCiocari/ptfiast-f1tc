Quais fatores parecem mais críticos para a satisfação?

Os fatores mais críticos são problemas com o suporte e atrasos de entrega. Problemas com suporte se refere tanto a quantidade de vezes que o cliente vem buscar o suporte, tempo de atendimento de um ticket, bem como reclamações externas. Fatores como idade do cliente, preço do produto, tempo de casa, e forma de pagamento não são tão relevantes para a nota NPS daquela jornada de compra.

O que mais gera detratores?

Atualmente, problemas com o suporte são o que mais afetam o NPS, gerando mais detratores. Vale destacar que, mesmo sem problemas de entrega, 36% dos clientes ainda viram detratores, mostrando o quão crítico é o problema com o suporte. Quando há atrasos de entrega, a partir do quarto dia de atraso 95% das notas viram detratoras.

Existe algum "ponto de ruptura" na experiência do cliente?

Existem quatro itens que direcionam o NPS mais fortemente, sendo três relacionados ao suporte (contato com suporte, dias de resolução de ticket, número de reclamações) e um de transporte (dias de atraso na entrega):

Um detrator típico experiencia 2 dias de atraso, sendo que promotores e neutros toleram apenas um dia, sendo que a partir do quarto dia de atraso, 95% dos clientes são detratores. 

Porém o que mais chama atenção é que, até uma reclamação o número de detratores não é alto (abaixo de 10%). Porém, a partir de duas reclamações, o número de detratores salta para 43%, e a partir de 3, dois terços dos clientes viram detratores. Falhar no atendimento na primeira reclamação é critico para o NPS.

O número de contatos com o suporte escala linearmente a medida que o número de reclamações aumenta. Reduzir o número de reclamações também impacta na carga operacional necessária para atender os contatos dos clientes.


Que tipo de cliente tende a ter NPS mais alto ou mais baixo?

Um cliente com pouco contato com o Suporte é quem tende a ter NPS mais alto. Mesmo recebendo dentro do prazo, clientes viram detratores caso venham a interagir com o suporte.

Um Promotor típico:

1 dia de atraso

1 contato com CS

3 dias de resolução de ticket

2 reclamações.

Já um cliente neutro típico:

1 dia de atraso

1 contato com CS

4 dias de resolução

3 reclamações

E um detrator típico:

2 dias de atraso

2 contatos com CS

6 dias de resolução de ticket

4 reclamações


O que demonstra que atrasos não são tolerados acima de um dia, e quanto mais contato com o suporte, seja interna ou externamente, o NPS vai piorando.
